**PART A: LOADING AND CONFIGURING A PRE-TRAINED LLM**

Perform the following tasks:

•	Select a suitable pre-trained LLM using Hugging Face Transformers.

•	Load the model and its tokenizer.

•	Prepare and tokenize sample user prompts.

•	Generate responses by configuring suitable parameters, such as temperature and maximum output length.

•	Display and examine the generated responses.


```python
# PART A: Loading & Configuring a Pre-Trained Transformer LLM

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
print("Sreenidhi S (24BAD114)\n")

# 1. Define model repository and load tokenizer & model
MODEL_ID = "Qwen/Qwen2.5-0.5B-Instruct"
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"[Roll No: 24BAD114] Loading {MODEL_ID} on {device}...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto" if torch.cuda.is_available() else None
)
if device == "cpu":
    model.to("cpu")

# 2. Prepare sample user prompt using chat template
prompt = "Explain what a neural network is in two simple sentences."
messages = [{"role": "user", "content": prompt}]
formatted_input = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

# 3. Tokenize input prompt
inputs = tokenizer(formatted_input, return_tensors="pt").to(device)

# 4. Generate response with configured temperature and max output length
with torch.no_grad():
    output_tokens = model.generate(
        **inputs,
        max_new_tokens=100,
        temperature=0.7,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

# 5. Decode and display the generated response
input_length = inputs["input_ids"].shape[1]
generated_tokens = output_tokens[0][input_length:]
response = tokenizer.decode(generated_tokens, skip_special_tokens=True)

print("\n--- Model Response ---")
print(response)
```

**PART B: IMPLEMENTING CHATBOT FUNCTIONALITY**

Perform the following tasks:

•	Create a function to accept user queries and generate responses using the loaded LLM.

•	Construct suitable prompts and display the generated responses.

•	Maintain conversation history, if supported by the model.

•	Test the chatbot with different queries and observe response quality.


```python
# PART B: Multi-Turn Chatbot with Conversation History

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_ID = "Qwen/Qwen2.5-0.5B-Instruct"
device = "cuda" if torch.cuda.is_available() else "cpu"

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto" if torch.cuda.is_available() else None
)
if device == "cpu":
    model.to("cpu")

class ConversationalChatbot:
    def __init__(self, model, tokenizer, device, system_prompt="You are a helpful and concise AI assistant."):
        self.model = model
        self.tokenizer = tokenizer
        self.device = device
        self.history = [{"role": "system", "content": system_prompt}]

    def chat(self, user_query, temperature=0.6, max_new_tokens=150):
        # Append user message to history
        self.history.append({"role": "user", "content": user_query})

        # Apply chat template
        prompt = self.tokenizer.apply_chat_template(
            self.history,
            tokenize=False,
            add_generation_prompt=True
        )
        inputs = self.tokenizer(prompt, return_tensors="pt").to(self.device)

        with torch.no_grad():
            output_tokens = self.model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                temperature=temperature,
                do_sample=True if temperature > 0 else False,
                pad_token_id=self.tokenizer.eos_token_id
            )

        input_len = inputs["input_ids"].shape[1]
        response_tokens = output_tokens[0][input_len:]
        reply = self.tokenizer.decode(response_tokens, skip_special_tokens=True).strip()

        # Append assistant reply to conversation history
        self.history.append({"role": "assistant", "content": reply})
        return reply

# Test the multi-turn session
bot = ConversationalChatbot(model, tokenizer, device)
print("[Roll No: 24BAD114] Testing multi-turn chatbot:\n")

queries = [
    "Hi, my name is Alex and I am studying AI.",
    "What is my name and what am I studying?",
    "Can you suggest two foundational project ideas for me?"
]

for q in queries:
    print(f"User: {q}")
    ans = bot.chat(q)
    print(f"Assistant: {ans}\n" + "-" * 40)
```

**PART C: DEVELOPING AN INTERACTIVE INTERFACE**

Perform the following tasks:

•	Select Streamlit or Gradio to develop the chatbot interface.

•	Create a text input component and a submission mechanism.

•	Connect the interface to the LLM response-generation function.

•	Display responses in a chat-style layout.

•	Run and test the application with different user inputs.


```python
# PART C: Interactive Web Interface using Gradio

import gradio as gr
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_ID = "Qwen/Qwen2.5-0.5B-Instruct"
device = "cuda" if torch.cuda.is_available() else "cpu"

print(f"[Roll No: 24BAD114] Initializing Gradio Chat Application...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto" if torch.cuda.is_available() else None
)
if device == "cpu":
    model.to("cpu")

def respond(message, history, system_prompt, temperature, max_new_tokens):
    # Construct message sequence
    messages = [{"role": "system", "content": system_prompt}]
    for user_turn, bot_turn in history:
        messages.append({"role": "user", "content": user_turn})
        messages.append({"role": "assistant", "content": bot_turn})
    messages.append({"role": "user", "content": message})

    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=int(max_new_tokens),
            temperature=float(temperature),
            do_sample=True if temperature > 0 else False,
            pad_token_id=tokenizer.eos_token_id
        )

    input_len = inputs["input_ids"].shape[1]
    reply = tokenizer.decode(output[0][input_len:], skip_special_tokens=True).strip()
    return reply

demo = gr.ChatInterface(
    fn=respond,
    title="Interactive LLM Chatbot [Roll No: 24BAD114]",
    description="A Transformer-based multi-turn chatbot interface powered by Hugging Face Transformers.",
    additional_inputs=[
        gr.Textbox(value="You are a helpful AI assistant.", label="System Prompt"),
        gr.Slider(minimum=0.1, maximum=1.5, value=0.7, step=0.1, label="Temperature"),
        gr.Slider(minimum=32, maximum=512, value=128, step=32, label="Max New Tokens")
    ],
    examples=[
        ["What is transfer learning in machine learning?"],
        ["Compare supervised and unsupervised learning in a small table."]
    ]
)

if __name__ == "__main__":
    demo.launch(share=True)
```
