```python
print("Sreenidhi S - 24BAD114")

# Step 1: Import libraries
import nltk
import string

from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Step 2: Download required resources
for resource in [
    "brown",
    "punkt",
    "punkt_tab",
    "stopwords",
    "wordnet",
    "omw-1.4"
]:
    nltk.download(resource, quiet=True)

print("\nAll required NLTK resources downloaded.")

# Step 3: Load Brown Corpus
sentences = brown.sents(categories="news")

print("\n1. DATASET PREPARATION")
print("Dataset: Brown Corpus")
print("Category: News")
print("Total sentences:", len(sentences))

# Step 4: Display sample sentences
print("\n2. SAMPLE SENTENCES")

for i, sentence in enumerate(sentences[:3], start=1):
    print(f"Sentence {i}:", " ".join(sentence))

# Step 5: Create original text from sample sentences
original_text = " ".join(
    " ".join(sentence) for sentence in sentences[:1000]
)

# Step 6: Convert text to lowercase
lowercase_text = original_text.lower()

# Step 7: Tokenize the text
tokens = word_tokenize(lowercase_text)

# Step 8: Remove punctuation and stop words
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word for word in tokens
    if word.isalpha() and word not in stop_words
]

# Step 9: Lemmatize tokens
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]

# Step 10: Generate preprocessed text
preprocessed_text = " ".join(lemmatized_tokens)

# Step 11: Display preprocessing results
print("\n3. ORIGINAL TEXT")
print(original_text[:400])

print("\n4. LOWERCASE TEXT")
print(lowercase_text[:300])

print("\n5. TOKENIZED WORDS")
print(tokens[:30])

print("\n6. AFTER PUNCTUATION AND STOP-WORD REMOVAL")
print(filtered_tokens[:30])

print("\n7. AFTER LEMMATIZATION")
print(lemmatized_tokens[:30])

print("\n8. PREPROCESSED TEXT")
print(preprocessed_text[:400])

# Step 12: Display statistics
print("\n9. PREPROCESSING STATISTICS")
print("Original token count:", len(tokens))
print("Tokens after filtering:", len(filtered_tokens))
print("Tokens after lemmatization:", len(lemmatized_tokens))
print("Unique words:", len(set(lemmatized_tokens)))
```

**PART B: VOCABULARY GENERATION AND WORD FREQUENCY ANALYSIS**

**Perform the following tasks:**

•	Create a vocabulary from the preprocessed text.

•	Calculate word frequencies using NLTK's FreqDist.

•	Identify the ten most frequently occurring words.

•	Display the vocabulary size and frequency distribution.

•	Prepare the tokenized sentences for Word2Vec training.



```python
# Step 1: Import required library
from nltk.probability import FreqDist
from nltk.corpus import brown, stopwords
from nltk.stem import WordNetLemmatizer

# Step 2: Generate vocabulary
vocabulary = set(lemmatized_tokens)

print("\n1. VOCABULARY")
print("Vocabulary size:", len(vocabulary))
print("Sample vocabulary words:", sorted(vocabulary)[:20])

# Step 3: Calculate word frequencies
frequency_distribution = FreqDist(lemmatized_tokens)

print("\n2. WORD FREQUENCY ANALYSIS")
print("Total tokens:", len(lemmatized_tokens))
print("Unique words:", len(frequency_distribution))

print("\nTop 10 most frequent words:")
for word, frequency in frequency_distribution.most_common(10):
    print(f"{word}: {frequency}")

# Step 4: Display frequency distribution
print("\n3. FREQUENCY DISTRIBUTION")
print(frequency_distribution)

# Step 5: Prepare tokenized sentences for Word2Vec
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

tokenized_sentences = []

for sentence in brown.sents(categories="news"):
    cleaned_sentence = []

    for word in sentence:
        word = word.lower()

        if word.isalpha() and word not in stop_words:
            cleaned_sentence.append(
                lemmatizer.lemmatize(word)
            )

    if cleaned_sentence:
        tokenized_sentences.append(cleaned_sentence)

print("\n4. TOKENIZED SENTENCES FOR WORD2VEC")
print("Total prepared sentences:", len(tokenized_sentences))
print("First prepared sentence:", tokenized_sentences[0])
print("Words in first sentence:", len(tokenized_sentences[0]))
```

**PART C: WORD EMBEDDING GENERATION USING WORD2VEC**

**Perform the following tasks:**

•	Train a Word2Vec model using Gensim on the preprocessed sentences.

•	Use the Skip-gram architecture with an embedding dimension of 100 and a context-window size of 5.

•	Generate embedding vectors for selected words.

•	Display the embedding vector and its dimensions.

•	Identify the five most similar words for a selected word using the trained model.



```python
# Step 1: Import required library
%pip install -q gensim

from gensim.models import Word2Vec

# Step 2: Train Word2Vec using Skip-gram
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=1,
    workers=2,
    epochs=10,
    seed=42
)

print("\n1. WORD2VEC MODEL TRAINED")
print("Model: Word2Vec")
print("Training method: Skip-gram")
print("Embedding dimensions:", model.vector_size)
print("Context window:", model.window)
print("Vocabulary size:", len(model.wv))

# Step 3: Display selected word vectors
print("\n2. WORD EMBEDDINGS")

selected_words = ["government", "people", "state"]

for word in selected_words:
    if word in model.wv:
        print(f"\nWord: {word}")
        print("Vector dimensions:", len(model.wv[word]))
        print("First 10 vector values:", model.wv[word][:10])
    else:
        print(f"\n'{word}' is not in the trained vocabulary.")

# Step 4: Find five similar words
print("\n3. FIVE MOST SIMILAR WORDS")

for word in selected_words:
    if word in model.wv:
        print(f"\nWords similar to '{word}':")
        for similar_word, score in model.wv.most_similar(
            word, topn=5
        ):
            print(f"{similar_word}: {score:.4f}")
    else:
        print(f"Cannot find similar words for '{word}'.")
```

**PART D: WORD SIMILARITY ANALYSIS AND EXPERIMENT MANAGEMENT**

**Perform the following tasks:**

•	Select three meaningful word pairs from the trained vocabulary.

•	Calculate the cosine similarity between each pair using the Word2Vec model.

•	Compare the similarity scores and identify the most similar pair.

•	Analyze the relationship between contextual similarity and the generated word embeddings.

•	Document the observations and upload the completed notebook to GitHub.


```python
# Step 1: Import required library
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import pandas as pd

# Step 2: Select three meaningful word pairs
word_pairs = [
    ("government", "state"),
    ("people", "man"),
    ("city", "country")
]

results = []

# Step 3: Calculate cosine similarity
for word1, word2 in word_pairs:
    if word1 in model.wv and word2 in model.wv:
        vector1 = model.wv[word1].reshape(1, -1)
        vector2 = model.wv[word2].reshape(1, -1)

        score = cosine_similarity(vector1, vector2)[0][0]

        results.append({
            "Word 1": word1,
            "Word 2": word2,
            "Cosine Similarity": round(float(score), 4)
        })

    else:
        print(
            f"Skipping pair ({word1}, {word2}): "
            "one or both words are not in the vocabulary."
        )

# Step 4: Display similarity results
print("\n1. COSINE SIMILARITY SCORES")

if results:
    results_df = pd.DataFrame(results)
    print(results_df.to_string(index=False))

    # Step 5: Identify the most similar pair
    best_pair = max(results, key=lambda item: item["Cosine Similarity"])

    print("\n2. MOST SIMILAR WORD PAIR")
    print("Word pair:", best_pair["Word 1"], "-", best_pair["Word 2"])
    print("Similarity score:", best_pair["Cosine Similarity"])

    # Step 6: Interpret the results
    print("\n3. OBSERVATIONS")

    for item in results:
        score = item["Cosine Similarity"]
        pair = f'{item["Word 1"]} - {item["Word 2"]}'

        if score >= 0.7:
            interpretation = "High similarity in the learned embedding space."
        elif score >= 0.4:
            interpretation = "Moderate similarity in the learned embedding space."
        elif score >= 0:
            interpretation = "Low positive similarity in the learned embedding space."
        else:
            interpretation = "Negative similarity in the learned embedding space."

        print(f"{pair}: {score:.4f} - {interpretation}")

else:
    print("No valid word pairs found.")
    print("Choose words available in model.wv and rerun Part D.")
```
