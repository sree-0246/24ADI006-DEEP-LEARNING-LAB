**Task A: VISUALIZATION OF ACTIVATION FUNCTIONS**

Implement and plot the following activation functions:

•	Sigmoid

•	Tanh

•	ReLU

Compare their:

•	Output range

•	Saturation regions

•	Gradient behavior

•	Computational efficiency

•	Typical applications



```python
# TASK A: VISUALIZATION OF ACTIVATION FUNCTIONS

print("SREENIDHI S - 24BAD114")
print("Task A: Visualization of Activation Functions")

# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1. Define Input Values
x = np.linspace(-10, 10, 400)

# 2. Define Activation Functions
# Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Tanh
def tanh(x):
    return np.tanh(x)

# ReLU
def relu(x):
    return np.maximum(0, x)

# Calculate outputs
y_sigmoid = sigmoid(x)
y_tanh = tanh(x)
y_relu = relu(x)

# 3. Plot Activation Functions
plt.figure(figsize=(12, 8))

plt.plot(x, y_sigmoid, label="Sigmoid", linewidth=2)
plt.plot(x, y_tanh, label="Tanh", linewidth=2)
plt.plot(x, y_relu, label="ReLU", linewidth=2)

plt.title("Comparison of Activation Functions")
plt.xlabel("Input (x)")
plt.ylabel("Activation Output")
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.grid(True)
plt.legend()

plt.show()

# 4. Individual Activation Function Plots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Sigmoid
axes[0].plot(x, y_sigmoid)
axes[0].set_title("Sigmoid")
axes[0].set_xlabel("Input")
axes[0].set_ylabel("Output")
axes[0].grid(True)

# Tanh
axes[1].plot(x, y_tanh)
axes[1].set_title("Tanh")
axes[1].set_xlabel("Input")
axes[1].set_ylabel("Output")
axes[1].grid(True)

# ReLU
axes[2].plot(x, y_relu)
axes[2].set_title("ReLU")
axes[2].set_xlabel("Input")
axes[2].set_ylabel("Output")
axes[2].grid(True)

plt.tight_layout()
plt.show()

# 5. Comparison Table
comparison = {
    "Activation Function": ["Sigmoid", "Tanh", "ReLU"],
    "Output Range": ["0 to 1", "-1 to 1", "0 to ∞"],
    "Saturation": [
        "Saturates at large positive/negative inputs",
        "Saturates at large positive/negative inputs",
        "No saturation for positive inputs"
    ],
    "Gradient Behavior": [
        "Can suffer from vanishing gradients",
        "Can suffer from vanishing gradients",
        "Better gradient flow for positive inputs"
    ],
    "Computational Efficiency": [
        "Moderate",
        "Moderate",
        "High"
    ],
    "Typical Applications": [
        "Binary classification output layer",
        "Hidden layers in some networks",
        "Hidden layers in deep neural networks"
    ]
}

df = pd.DataFrame(comparison)

print("\nACTIVATION FUNCTION COMPARISON")
print("=" * 60)

display(df)
```

**Task B: PERFORMANCE COMPARISON OF ACTIVATION FUNCTIONS**

Train identical ANN architectures using:

•	Sigmoid

•	Tanh

•	ReLU

Record and compare:

•	Training accuracy

•	Validation accuracy

•	Training loss

•	Validation loss

•	Number of epochs required for convergence

Analyze which activation function performs best and justify the observations.


```python
print("SREENIDHI S - 24BAD114")
print("Task B: Performance Comparison of Activation Functions")

# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

# Import TensorFlow components
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# Import dataset and preprocessing tools
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_classification

# Set random seeds for reproducible results
np.random.seed(42)
tf.random.set_seed(42)

# Create a binary classification dataset
X, y = make_classification(
    n_samples=2000,
    n_features=10,
    n_informative=8,
    n_redundant=2,
    n_classes=2,
    random_state=42
)

# Split the dataset into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Standardize the input features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)

# Display dataset information
print("\nDataset Information")
print("Total Samples       :", len(X))
print("Training Samples    :", len(X_train))
print("Validation Samples  :", len(X_val))
print("Number of Features  :", X_train.shape[1])

# Function to create the same ANN architecture
def create_model(activation):

    # Create sequential neural network
    model = Sequential([
        # First hidden layer
        Dense(32, activation=activation, input_shape=(10,)),

        # Second hidden layer
        Dense(16, activation=activation),

        # Output layer for binary classification
        Dense(1, activation="sigmoid")
    ])

    # Compile the model using Adam optimizer
    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model

# Define activation functions for comparison
activation_functions = ["sigmoid", "tanh", "relu"]

# Store training histories and results
histories = {}
results = {}

# Set training parameters
EPOCHS = 30
BATCH_SIZE = 32

# Train the ANN with each activation function
for activation in activation_functions:

    print("\nTraining using", activation.upper())

    # Create the ANN with the selected activation function
    model = create_model(activation)

    # Train the model
    history = model.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=0
    )

    # Store the training history
    histories[activation] = history.history

    # Get final training and validation accuracy
    training_accuracy = history.history["accuracy"][-1]
    validation_accuracy = history.history["val_accuracy"][-1]

    # Get final training and validation loss
    training_loss = history.history["loss"][-1]
    validation_loss = history.history["val_loss"][-1]

    # Find the epoch with the highest validation accuracy
    convergence_epoch = np.argmax(
        history.history["val_accuracy"]
    ) + 1

    # Store all results
    results[activation] = {
        "Training Accuracy": training_accuracy,
        "Validation Accuracy": validation_accuracy,
        "Training Loss": training_loss,
        "Validation Loss": validation_loss,
        "Convergence Epoch": convergence_epoch
    }

# Convert results into a DataFrame
results_df = pd.DataFrame(results).T

# Display performance comparison
print("\nPERFORMANCE COMPARISON")
print("=" * 60)
print(results_df.round(4))

# Plot training loss for all activation functions
plt.figure(figsize=(10, 6))

for activation in activation_functions:
    plt.plot(
        histories[activation]["loss"],
        label=activation.upper()
    )

plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.legend()
plt.grid(True)
plt.show()

# Plot validation loss for all activation functions
plt.figure(figsize=(10, 6))

for activation in activation_functions:
    plt.plot(
        histories[activation]["val_loss"],
        label=activation.upper()
    )

plt.title("Validation Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)
plt.show()

# Plot training accuracy for all activation functions
plt.figure(figsize=(10, 6))

for activation in activation_functions:
    plt.plot(
        histories[activation]["accuracy"],
        label=activation.upper()
    )

plt.title("Training Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Plot validation accuracy for all activation functions
plt.figure(figsize=(10, 6))

for activation in activation_functions:
    plt.plot(
        histories[activation]["val_accuracy"],
        label=activation.upper()
    )

plt.title("Validation Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Find the activation function with the highest validation accuracy
best_activation = max(
    results,
    key=lambda x: results[x]["Validation Accuracy"]
)

# Display the best performing activation function
print("\nBEST PERFORMING ACTIVATION FUNCTION")
print("=" * 60)

print(
    "Best Activation Function :",
    best_activation.upper()
)

print(
    "Training Accuracy        :",
    round(results[best_activation]["Training Accuracy"], 4)
)

print(
    "Validation Accuracy      :",
    round(results[best_activation]["Validation Accuracy"], 4)
)

print(
    "Training Loss            :",
    round(results[best_activation]["Training Loss"], 4)
)

print(
    "Validation Loss          :",
    round(results[best_activation]["Validation Loss"], 4)
)

print(
    "Convergence Epoch        :",
    results[best_activation]["Convergence Epoch"]
)

# Display the analysis
print("\nANALYSIS")

print(
    best_activation.upper(),
    "achieved the highest validation accuracy in this experiment."
)

print(
    "The same ANN architecture, dataset, optimizer, batch size "
    "and epochs were used for all three activation functions."
)

print(
    "Higher validation accuracy and lower validation loss "
    "indicate better performance."
)
```

**Task C: COMPARISON OF OPTIMIZATION ALGORITHMS**

Train the same ANN using:

•	Stochastic Gradient Descent (SGD)

•	Momentum

•	RMSProp

•	Adam

Compare:

•	Training loss

•	Validation loss

•	Convergence speed

•	Final training accuracy

•	Final validation accuracy

Visualize:

•	Loss vs Epoch

•	Accuracy vs Epoch

Discuss which optimizer converges faster and produces better generalization.


```python
print("SREENIDHI S - 24BAD114")
print("Task C: Comparison of Optimization Algorithms")

# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

# Import optimizers
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD, RMSprop, Adam

# Import dataset and preprocessing tools
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_classification

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# Create a binary classification dataset
X, y = make_classification(
    n_samples=2000,
    n_features=10,
    n_informative=8,
    n_redundant=2,
    n_classes=2,
    random_state=42
)

# Split dataset into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Standardize the input features
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)

# Display dataset information
print("\nDataset Information")
print("Total Samples       :", len(X))
print("Training Samples    :", len(X_train))
print("Validation Samples  :", len(X_val))
print("Number of Features  :", X_train.shape[1])

# Function to create the same ANN architecture
def create_model(optimizer):

    # Create ANN with two hidden layers
    model = Sequential([
        Dense(32, activation="relu", input_shape=(10,)),
        Dense(16, activation="relu"),
        Dense(1, activation="sigmoid")
    ])

    # Compile model using selected optimizer
    model.compile(
        optimizer=optimizer,
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model

# Define optimization algorithms
optimizers = {
    "SGD": SGD(learning_rate=0.01),
    "Momentum": SGD(learning_rate=0.01, momentum=0.9),
    "RMSProp": RMSprop(learning_rate=0.001),
    "Adam": Adam(learning_rate=0.001)
}

# Store training histories and results
histories = {}
results = {}

# Set training parameters
EPOCHS = 30
BATCH_SIZE = 32

# Train the same ANN using each optimizer
for name, optimizer in optimizers.items():

    print("\nTraining using", name)

    # Create a fresh ANN
    model = create_model(optimizer)

    # Train the model
    history = model.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=0
    )

    # Store training history
    histories[name] = history.history

    # Get final training loss
    training_loss = history.history["loss"][-1]

    # Get final validation loss
    validation_loss = history.history["val_loss"][-1]

    # Get final training accuracy
    training_accuracy = history.history["accuracy"][-1]

    # Get final validation accuracy
    validation_accuracy = history.history["val_accuracy"][-1]

    # Find epoch with highest validation accuracy
    convergence_epoch = (
        np.argmax(history.history["val_accuracy"]) + 1
    )

    # Store results
    results[name] = {
        "Training Loss": training_loss,
        "Validation Loss": validation_loss,
        "Convergence Epoch": convergence_epoch,
        "Training Accuracy": training_accuracy,
        "Validation Accuracy": validation_accuracy
    }

# Convert results into a DataFrame
results_df = pd.DataFrame(results).T

# Display comparison table
print("\nOPTIMIZATION ALGORITHM COMPARISON")
print("=" * 60)
print(results_df.round(4).to_string())

# Plot training and validation loss
plt.figure(figsize=(10, 6))

for name in optimizers:
    plt.plot(
        histories[name]["loss"],
        label=name + " Training"
    )

    plt.plot(
        histories[name]["val_loss"],
        linestyle="--",
        label=name + " Validation"
    )

plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

# Plot training and validation accuracy
plt.figure(figsize=(10, 6))

for name in optimizers:
    plt.plot(
        histories[name]["accuracy"],
        label=name + " Training"
    )

    plt.plot(
        histories[name]["val_accuracy"],
        linestyle="--",
        label=name + " Validation"
    )

plt.title("Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Find optimizer with highest validation accuracy
best_optimizer = max(
    results,
    key=lambda x: results[x]["Validation Accuracy"]
)

# Find fastest converging optimizer
fastest_optimizer = min(
    results,
    key=lambda x: results[x]["Convergence Epoch"]
)

# Display best optimizer results
print("\nBEST OPTIMIZER FOR GENERALIZATION")
print("=" * 60)

print(
    "Best Optimizer            :",
    best_optimizer
)

print(
    "Final Training Accuracy   :",
    round(
        results[best_optimizer]["Training Accuracy"],
        4
    )
)

print(
    "Final Validation Accuracy :",
    round(
        results[best_optimizer]["Validation Accuracy"],
        4
    )
)

print(
    "Final Training Loss       :",
    round(
        results[best_optimizer]["Training Loss"],
        4
    )
)

print(
    "Final Validation Loss     :",
    round(
        results[best_optimizer]["Validation Loss"],
        4
    )
)

print(
    "Convergence Epoch         :",
    results[best_optimizer]["Convergence Epoch"]
)

# Display fastest optimizer
print("\nFASTEST CONVERGING OPTIMIZER")
print("=" * 60)

print(
    "Fastest Optimizer         :",
    fastest_optimizer
)

print(
    "Convergence Epoch         :",
    results[fastest_optimizer]["Convergence Epoch"]
)

# Display analysis
print("\nANALYSIS")
print("=" * 60)

print(
    best_optimizer,
    "provided the best generalization based on validation accuracy."
)

print(
    fastest_optimizer,
    "converged fastest based on the convergence epoch."
)

print(
    "SGD uses basic gradient descent, while Momentum improves SGD "
    "by reducing oscillations."
)

print(
    "RMSProp adapts the learning rate for individual parameters, "
    "while Adam combines adaptive learning rates with momentum."
)

print(
    "Higher validation accuracy and lower validation loss indicate "
    "better generalization."
)
```
