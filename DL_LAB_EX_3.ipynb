```python
print("SREENIDHI S - 24BAD114")
print("Experiment 3: Implementation of Convolutional Neural Networks (CNNs) for Image Classification")

# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.optimizers import Adam

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Set random seeds for reproducible results
np.random.seed(42)
tf.random.set_seed(42)

# Define CIFAR-10 class names
class_names = [
    "airplane",
    "automobile",
    "bird",
    "cat",
    "deer",
    "dog",
    "frog",
    "horse",
    "ship",
    "truck"
]

# Task A: Load CIFAR-10 dataset
print("\nTask A: Dataset Preparation")
print("=" * 60)

print("Loading CIFAR-10 dataset...")

(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Display original dataset shape
print("\nOriginal Dataset")
print("Training Images :", X_train.shape)
print("Training Labels :", y_train.shape)
print("Testing Images  :", X_test.shape)
print("Testing Labels  :", y_test.shape)

# Normalize image pixel values
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Create validation set
validation_size = 5000

X_val = X_train[-validation_size:]
y_val = y_train[-validation_size:]

X_train = X_train[:-validation_size]
y_train = y_train[:-validation_size]

# Display processed dataset information
print("\nAfter Preprocessing")
print("Training Images   :", X_train.shape)
print("Validation Images :", X_val.shape)
print("Testing Images    :", X_test.shape)
print("Image Shape       :", X_train.shape[1:])
print("Number of Classes :", len(class_names))
print("Pixel Range       :", X_train.min(), "to", X_train.max())

# Display sample images
plt.figure(figsize=(10, 6))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i])
    plt.title(class_names[y_train[i][0]])
    plt.axis("off")

plt.suptitle("Sample CIFAR-10 Images")
plt.tight_layout()
plt.show()

# Task B: Build CNN model
print("\nTask B: CNN Model Implementation")
print("=" * 60)

# Create CNN architecture
model = Sequential([
    Conv2D(32, (3, 3), activation="relu", input_shape=(32, 32, 3)),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),

    Conv2D(128, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation="relu"),

    Dense(10, activation="softmax")
])

# Display CNN architecture
print("\nCNN Model Summary")
model.summary()

# Compile the CNN
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nModel compiled successfully.")

# Task C: Train and evaluate the CNN
print("\nTask C: Model Training and Evaluation")
print("=" * 60)

# Set training parameters
EPOCHS = 10
BATCH_SIZE = 64

# Train the model
history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    verbose=1
)

# Extract training results
training_accuracy = history.history["accuracy"]
validation_accuracy = history.history["val_accuracy"]

training_loss = history.history["loss"]
validation_loss = history.history["val_loss"]

# Evaluate the model on test data
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

# Display final results
print("\nFinal Performance")
print("=" * 60)

print(
    "Final Training Accuracy   :",
    round(training_accuracy[-1] * 100, 2),
    "%"
)

print(
    "Final Validation Accuracy :",
    round(validation_accuracy[-1] * 100, 2),
    "%"
)

print(
    "Final Training Loss       :",
    round(training_loss[-1], 4)
)

print(
    "Final Validation Loss     :",
    round(validation_loss[-1], 4)
)

print(
    "Testing Accuracy          :",
    round(test_accuracy * 100, 2),
    "%"
)

# Plot accuracy vs epoch
plt.figure(figsize=(10, 6))

plt.plot(
    training_accuracy,
    label="Training Accuracy"
)

plt.plot(
    validation_accuracy,
    label="Validation Accuracy"
)

plt.title("Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Plot loss vs epoch
plt.figure(figsize=(10, 6))

plt.plot(
    training_loss,
    label="Training Loss"
)

plt.plot(
    validation_loss,
    label="Validation Loss"
)

plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

# Task D: Performance Analysis
print("\nTask D: Performance Analysis")
print("=" * 60)

# Generate predictions
predictions = model.predict(
    X_test,
    verbose=0
)

# Convert probabilities to predicted class numbers
predicted_classes = np.argmax(
    predictions,
    axis=1
)

actual_classes = y_test.flatten()

# Calculate classification accuracy
classification_accuracy = np.mean(
    predicted_classes == actual_classes
)

print(
    "\nClassification Accuracy:",
    round(classification_accuracy * 100, 2),
    "%"
)

# Generate confusion matrix
cm = confusion_matrix(
    actual_classes,
    predicted_classes
)

# Display confusion matrix
plt.figure(figsize=(10, 8))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

disp.plot(
    xticks_rotation=45,
    values_format="d"
)

plt.title("CIFAR-10 Confusion Matrix")
plt.tight_layout()
plt.show()

# Display sample predictions
plt.figure(figsize=(12, 8))

for i in range(12):
    plt.subplot(3, 4, i + 1)

    plt.imshow(X_test[i])

    predicted_label = class_names[predicted_classes[i]]
    actual_label = class_names[actual_classes[i]]

    plt.title(
        "Predicted: " + predicted_label +
        "\nActual: " + actual_label
    )

    plt.axis("off")

plt.suptitle("Sample Predictions")
plt.tight_layout()
plt.show()

# Find misclassified images
misclassified_indices = np.where(
    predicted_classes != actual_classes
)[0]

print(
    "\nNumber of Misclassified Images:",
    len(misclassified_indices)
)

# Display misclassified images
plt.figure(figsize=(12, 8))

number_of_images = min(
    12,
    len(misclassified_indices)
)

for i in range(number_of_images):

    index = misclassified_indices[i]

    plt.subplot(3, 4, i + 1)

    plt.imshow(X_test[index])

    predicted_label = class_names[predicted_classes[index]]
    actual_label = class_names[actual_classes[index]]

    plt.title(
        "Predicted: " + predicted_label +
        "\nActual: " + actual_label
    )

    plt.axis("off")

plt.suptitle("Misclassified Images")
plt.tight_layout()
plt.show()

# Display CNN strengths
print("\nStrengths of CNNs")
print("=" * 60)

print("1. Automatically extracts important image features.")
print("2. Preserves spatial relationships between image pixels.")
print("3. Uses fewer parameters than fully connected networks.")
print("4. Provides effective performance for image classification.")
print("5. Can learn hierarchical features such as edges, shapes and objects.")

# Display CNN limitations
print("\nLimitations of CNNs")
print("=" * 60)

print("1. CNNs may require large amounts of training data.")
print("2. Training can require significant computational resources.")
print("3. CNNs can suffer from overfitting.")
print("4. Choosing suitable hyperparameters can be challenging.")
print("5. Training time can increase for deeper CNN architectures.")

# Display final experiment summary
print("\nExperiment Summary")
print("=" * 60)

print("Dataset                : CIFAR-10")
print("Training Images        :", len(X_train))
print("Validation Images      :", len(X_val))
print("Testing Images         :", len(X_test))
print("CNN Architecture       : Conv2D + ReLU + MaxPooling + Dense + Softmax")
print("Number of Epochs       :", EPOCHS)
print("Batch Size             :", BATCH_SIZE)
print("Optimizer              : Adam")
print("Loss Function          : Sparse Categorical Cross-Entropy")
print(
    "Testing Accuracy       :",
    round(test_accuracy * 100, 2),
    "%"
)

print("\nExperiment completed successfully.")
```
